# Plot permutation test results

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import json
```

```python
with open('shuffle_label/permutation_test_label_10000targets_3000cpgs.json', 'r') as f:
    label_data = json.load(f)

with open('shuffle_target/permutation_test_target_permutation_data.json', 'r') as f:
    target_data = json.load(f)
```

```python
fig, ax = plt.subplots(figsize=(11, 4.5), dpi=300, ncols=2)

ax[0].hist(label_data['random_percentages'], bins=np.arange(0, 100 + 5, 5), edgecolor='black')
ax[0].axvline(x=label_data['observed_percentage'], color='red', linestyle='dashed', linewidth=2)
ax[0].set_title("Random Shuffle Tissue Labels\n", fontsize=14)
ax[0].text(62, 300, f"P-value <0.001\nObserved={label_data['observed_percentage']:.1f}%", fontsize=11)
ax[0].set_ylabel("Frequency\n", fontsize=12)

ax[1].hist(target_data['random_percentages'], bins=np.arange(0, 100 + 5, 5), edgecolor='black', color='tab:orange')
ax[1].axvline(x=target_data['observed_percentage'], color='red', linestyle='dashed', linewidth=2)
ax[1].set_title("Random Shuffle Microsatellite Genotypes\n", fontsize=14)
ax[1].text(62, 28, f"P-value <0.01\nObserved={target_data['observed_percentage']:.1f}%", fontsize=11)
ax[1].set_ylabel("\nFrequency\n", fontsize=12)

for ax in ax:
    ax.set_xlabel("\nPercent of Same-Tissue Cell Pairs", fontsize=12)
    ax.set_xlim(0, 100)
    ax.spines['right'].set_visible(False)
    ax.spines['top'].set_visible(False)
    ax.set_xlim(0, 100)
    ax.set_xticks(np.arange(0, 101, 10))

plt.tight_layout()
plt.savefig('permutation_test_combined_plot.pdf', bbox_inches='tight')
plt.show()
```

```python
# Individual plot 1: Random Shuffle Tissue Labels
plt.figure(figsize=(5.5, 4), dpi=300)

plt.hist(label_data['random_percentages'], bins=np.arange(0, 100 + 5, 5), edgecolor='black')
plt.axvline(x=label_data['observed_percentage'], color='red', linestyle='dashed', linewidth=2)
plt.title("Random Shuffle Tissue Labels\n", fontsize=14)
plt.text(62, 300, f"P-value <0.001\nObserved={label_data['observed_percentage']:.1f}%", fontsize=11)
plt.ylabel("Frequency\n", fontsize=12)
plt.xlabel("\nPercent of Same-Tissue Cell Pairs", fontsize=12)
plt.xlim(0, 100)
plt.xticks(np.arange(0, 101, 10))

ax = plt.gca()
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)

plt.tight_layout()
plt.savefig('permutation_test_shuffle_labels.pdf', bbox_inches='tight')
plt.show()
```

```python
# Individual plot 2: Random Shuffle Microsatellite Genotypes
plt.figure(figsize=(5.5, 4), dpi=300)

plt.hist(target_data['random_percentages'], bins=np.arange(0, 100 + 5, 5), edgecolor='black', color='tab:orange')
plt.axvline(x=target_data['observed_percentage'], color='red', linestyle='dashed', linewidth=2)
plt.title("Random Shuffle Microsatellite Genotypes\n", fontsize=14)
plt.text(62, 28, f"P-value <0.01\nObserved={target_data['observed_percentage']:.1f}%", fontsize=11)
plt.ylabel("Frequency\n", fontsize=12)
plt.xlabel("\nPercent of Same-Tissue Cell Pairs", fontsize=12)
plt.xlim(0, 100)
plt.xticks(np.arange(0, 101, 10))

ax = plt.gca()
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)

plt.tight_layout()
plt.savefig('permutation_test_shuffle_targets.pdf', bbox_inches='tight')
plt.show()
```
